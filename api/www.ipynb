枚举 Enum

```python
from enum import Enum
from typing import Union


# 定义不继承自 str 的枚举类
class FilterType(str, Enum):
    user_os = "userOs"
    user_browser = "userBrowser"
    user_device = "userDevice"


# 测试函数
def compare_enum_and_string(abcw: Union[FilterType, str]):
    if abcw in [FilterType.user_device]:
        print(111)
    print([FilterType.user_device])
    print(abcw, type(abcw))


# 测试
compare_enum_and_string(FilterType.user_device)  # 输出：Enum member matched
compare_enum_and_string("userDevice")  # 输出：String matched
```

```python
from enum import Enum


# 定义 FetchFilterType 枚举类
class FetchFilterType(Enum):
    _url = "url"
    _name = "name"
    _api = "api"


# 定义 EventType 枚举类
class EventType(Enum):
    request = "request"
    graphql = "graphql"
    location = "location"


# 定义一个函数来判断并转换类型
def check_and_convert(type: FetchFilterType):
    # 使用 in 语法判断 type 是否在某个列表中
    if type in [FetchFilterType._url, FetchFilterType._api]:
        new_type = EventType.request
    elif type in [FetchFilterType._name]:
        new_type = EventType.graphql
    else:
        new_type = EventType.location

    print(f"原始类型: {type}, 转换后的类型: {new_type}")


check_and_convert(FetchFilterType._url)  # 输出: 原始类型: FetchFilterType._url, 转换后的类型: EventType.request


check_and_convert(FetchFilterType._name)  # 输出: 原始类型: FetchFilterType._name, 转换后的类型: EventType.graphql
check_and_convert(FetchFilterType._api)  # 输出: 原始类型: FetchFilterType._api, 转换后的类型: EventType.request
```
